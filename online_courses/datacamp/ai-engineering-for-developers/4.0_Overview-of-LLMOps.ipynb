#### **4.2 Development Phase**

![alt text](<Screenshot 2026-06-07 011603.png>)


- Prompt engineering
- Chains & Agents:
    - Chains: A pipeline connects steps that take inputs & produce outputs
    - Agents: multiple actions, LLM decide which action, produce output with reinforce info
- Retrieval-Augmented Generation (RAG) vs Fine-tuning:
    - RAG (Factual Knowledge): 
        1. Convert inputs to embedding (pre-trained model convert text to numerical representation). [R]
        2. Search vector databases containing all embeddings. Compare & calcualte similarity [R]
        3. Retrieve most similar documents. [R]
        4. Combine input of combined documents to create final prompt. [A]
        5. Uses prompt to create an output. [G]
    - Fine-tuning (Specialized New Domain):
        - Supervised fine-tuning (transfer learning)
        - Reinforcement Learning from Human Feedback
- Testing
    1. Building a test set: closely resemble real-world scenarios.
    2. Choose the right metric:
        - Right or Wrong answer?: Accuracy
        - Reference Answer?: Text comparison metrics (statistical / LLM judges)
        - Human feedback?: Feedback score metrics / Model-based / LLM judges
        - No human feedback?: Unsupervised metrics (Coherence / Fluency / Diversity )
    3. Define optional secondary metrics:
        - Output characteristis (bias/toxicity/helpfulness)
        - Operational characteristis (latency/cost/memory usage)

#### **4.3 Operational Phase**
- Deployment:
    1. Choice of hosting:
        - Private/Public cloud or On-premise hosting
    2. API design
    3. How to run using:
        - Containers
        - Serverless function
        - Cloud managed services
    4. Continous Integration (CI) / Continous Deployment (CD)
        - CI:
            1. Source: Retrieve source code
            2. Build: Create a container image containing code
            3. Test: Perform integration tests
            4. Register: Store container in a registry
        - CD:
            1. Retrieve: Retrieve container from registry
            2. Test: Perform deployment tests
            3. Deploy: Deploy container to environments:
                - Staging
                - Production
- Monitoring (input, function, output) & Observability
- Cost Management:
    - Choose the right model
    - Optimize prompts
    - Optimize number of calls
- Governance & Security
    - Prompt Injection
    - Data Poisoning
    - Denial of Service
