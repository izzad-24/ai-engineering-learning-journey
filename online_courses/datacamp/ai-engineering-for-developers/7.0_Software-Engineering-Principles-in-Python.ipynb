**Inheritance in Python**

![alt text](<Screenshot 2026-07-31 121042.png>)

````python
from .parent_class import ParentClass

# to create a child class with inheritance from ParentClass
class Childclass(ParentClass):
    def __init__(self):
        # call parent's __init __ method
        ParentClass.__init__(self)
        # add attribute unique to child class
        self.child_attribute = "I am a child class attribute"

# calling it out
child_class = Childclass()
print(child_class.parent_attribute)
print(child_class.child_attribute)
````

### Multilevel Inheritance


````python
class ParentClass:
    def __init__(self):
        print("Parent class constructor")


class ChildClass(ParentClass):
    def __init__(self):
        ParentClass.__init__(self)
        print("Child class constructor")


class SuperChild(ChildClass):
    def __init__(self):
        super().__init__()
        print("Super child class constructor")


class AnotherSuperChild(ChildClass):
    def __init__(self):
        super().__init__()
        print("Super another child class constructor")


grandChild = AnotherSuperChild() # create instances of the classes
````
